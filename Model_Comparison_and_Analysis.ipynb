Model Comparison and Analysis

This notebook compares the performance of the trained machine learning models:
- Decision Tree
- Random Forest
- XGBoost

The comparison is based on:
- Accuracy
- Precision
- Recall
- F1-score
- Training time
- Prediction time

```python
from google.colab import drive
drive.mount('/content/drive')
```

Import libraries

```python
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
```

Set paths for the Dataset and Results

```python
PROJECT_DIR = Path("/content/drive/MyDrive/MSc_Dissertation")

PROCESSED_DIR = PROJECT_DIR / "Dataset/Processed"

RESULTS_DIR = PROJECT_DIR / "Results/Models"

decision_tree_results = pd.read_csv(
    RESULTS_DIR / "decision_tree_results.csv"
)

random_forest_results = pd.read_csv(
    RESULTS_DIR / "random_forest_results.csv"
)

xgboost_results = pd.read_csv(
    RESULTS_DIR / "xgboost_results.csv"
)
```

Combine results

```python
results = pd.concat(
    [
        decision_tree_results,
        random_forest_results,
        xgboost_results
    ],
    ignore_index=True
)


results
```

Save the comparision table

```python
results.to_csv(
    PROCESSED_DIR / "model_comparison_results.csv",
    index=False
)
```

Accuracy comparision

```python
plt.figure(figsize=(8,5))

plt.bar(
    results["Model"],
    results["Accuracy"]
)

plt.ylabel("Accuracy")

plt.title(
    "Model Accuracy Comparison"
)

plt.ylim(0.99,1.0)

plt.show()
```

F1 score comparision

```python
plt.figure(figsize=(8,5))

plt.bar(
    results["Model"],
    results["F1-score"]
)

plt.ylabel("F1-score")

plt.title(
    "Model F1-score Comparison"
)

plt.ylim(0.99,1.0)

plt.show()
```

Training time comparision

```python
plt.figure(figsize=(8,5))

plt.bar(
    results["Model"],
    results["Training Time (seconds)"]
)

plt.ylabel("Training Time (seconds)")

plt.title(
    "Model Training Time Comparison"
)

plt.show()
```
